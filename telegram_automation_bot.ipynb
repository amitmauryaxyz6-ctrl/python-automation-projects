```python

import requests
from datetime import datetime

# 1. Telegram Bot Configuration (Replace with real token & chat ID in production)
BOT_TOKEN = "YOUR_TELEGRAM_BOT_TOKEN"
CHAT_ID = "YOUR_CHANNEL_OR_USER_ID"

def fetch_live_market_data():
    """Fetch live Bitcoin price from public API to simulate real-time alert data."""
    domain = "api.coingecko" + ".com"
    url = f"https://{domain}/api/v3/simple/price?ids=bitcoin,ethereum&vs_currencies=inr,usd"
    response = requests.get(url)
    return response.json()

def send_telegram_alert(message_text):
    """Send formatted message to Telegram via official Bot API."""
    tg_domain = "api.telegram" + ".org"
    send_url = f"https://{tg_domain}/bot{BOT_TOKEN}/sendMessage"

    payload = {
        "chat_id": CHAT_ID,
        "text": message_text,
        "parse_mode": "Markdown"
    }

    # Preview the automated message before sending
    print("--- AUTOMATED TELEGRAM MESSAGE PREVIEW ---")
    print(message_text)
    print("-" * 45)

    if BOT_TOKEN == "YOUR_TELEGRAM_BOT_TOKEN":
        print("Status: Demo Mode (Add your BotFather Token to send live to Telegram).")
    else:
        res = requests.post(send_url, data=payload)
        print("Telegram API Response:", res.json())

# 2. Execute workflow
data = fetch_live_market_data()
current_time = datetime.now().strftime("%d-%b-%Y %I:%M %p")

alert_message = f"""🚀 *Automated Market Update*
🕒 *Time:* {current_time}

💰 *Bitcoin (BTC):* ₹{data['bitcoin']['inr']:,} (${data['bitcoin']['usd']:,})
💎 *Ethereum (ETH):* ₹{data['ethereum']['inr']:,} (${data['ethereum']['usd']:,})

_Sent automatically via Python Telegram Bot_"""

send_telegram_alert(alert_message)
```
