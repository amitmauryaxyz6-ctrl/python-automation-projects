```python

import requests
from bs4 import BeautifulSoup
import pandas as pd

# लिंक को जोड़कर बनाना ताकि मोबाइल ब्राउज़र इसे काट न सके
domain = "books" + ".toscrape.com"
url = "https://" + domain + "/"

response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

products = soup.find_all("article", class_="product_pod")
product_list = []

for item in products:
    name = item.h3.a["title"]
    price_text = item.find("p", class_="price_color").text
    clean_price = "".join(c for c in price_text if c.isdigit() or c == ".")
    price = float(clean_price)
    stock = item.find("p", class_="instock availability").text.strip()

    product_list.append({
        "Product_Name": name,
        "Price": price,
        "Availability": stock
    })

if len(product_list) > 0:
    df = pd.DataFrame(product_list)
    df.to_csv("ecommerce_prices.csv", index=False)
    cheap_deals = df[df["Price"] < 20.0]
    print(f"बधाई हो! कुल {len(df)} प्रोडक्ट्स का डेटा निकाला गया!\n")
    print("सबसे सस्ती डील्स (20 से कम कीमत वाले):")
    print(cheap_deals.head())
else:
    print("वेबसाइट से डेटा नहीं मिल पाया, कृपया इंटरनेट चेक करें।")
```
