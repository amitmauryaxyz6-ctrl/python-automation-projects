```python

import pandas as pd

# 1. Job requirements set by HR
required_skills = ["python", "pandas", "web scraping", "excel", "sql"]
job_role = "Python Automation Developer"

# 2. Sample candidate resumes data
candidates = [
    {"Name": "Aarav Sharma", "Email": "aarav@example.com", "Resume_Text": "Experienced in Python, SQL, and Excel data analysis."},
    {"Name": "Neha Verma", "Email": "neha@example.com", "Resume_Text": "Graphic designer skilled in Photoshop, Illustrator, and Canva."},
    {"Name": "Rohan Gupta", "Email": "rohan@example.com", "Resume_Text": "Python developer with strong skills in Pandas, Web Scraping, SQL, and Excel."},
    {"Name": "Simran Kaur", "Email": "simran@example.com", "Resume_Text": "Content writer and digital marketer with SEO knowledge."},
    {"Name": "Kabir Singh", "Email": "kabir@example.com", "Resume_Text": "Junior coder learning Python and basic Excel."}
]

# 3. Analyze resumes and calculate match score
results = []

for person in candidates:
    text = person["Resume_Text"].lower()
    matched = [skill.title() for skill in required_skills if skill in text]
    score = round((len(matched) / len(required_skills)) * 100)
    status = "Shortlisted" if score >= 60 else "Rejected"

    results.append({
        "Candidate_Name": person["Name"],
        "Email": person["Email"],
        "Match_Score (%)": score,
        "Matched_Skills": ", ".join(matched) if matched else "None",
        "HR_Status": status
    })

# 4. Sort by highest score and save shortlisted report to Excel (CSV)
df = pd.DataFrame(results).sort_values(by="Match_Score (%)", ascending=False)
df.to_csv("hr_shortlisted_candidates.csv", index=False)

print(f"--- HR RESUME SCREENING REPORT: {job_role} ---\n")
print(df.to_string(index=False))
print("\nSuccess! Shortlisted report saved to 'hr_shortlisted_candidates.csv'.")
```
