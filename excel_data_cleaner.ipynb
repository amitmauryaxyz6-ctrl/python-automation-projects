```python

import pandas as pd

# 1. Create sample messy client data (Raw Data)
raw_data = {
    "Customer_Name": ["  rahul sharma ", "Anita Verma", "rahul sharma", "Vikram Singh  ", None, "Priya Patel"],
    "Email": ["RAHUL@GMAIL.COM", "anita@yahoo.com", "rahul@gmail.com", "vikram@outlook.com", "unknown@test.com", "PRIYA@GMAIL.COM"],
    "City": ["lucknow", "Delhi", "lucknow", None, "Mumbai", "delhi"],
    "Purchase_Amount": [1500, 2500, 1500, 3200, 1200, 4100]
}

df_dirty = pd.DataFrame(raw_data)
print("--- BEFORE CLEANING (Messy Data) ---")
print(df_dirty)
print("\n")

# 2. Clean and standardize the data
df_clean = df_dirty.copy()

# Remove rows where Customer Name is missing
df_clean = df_clean.dropna(subset=["Customer_Name"])

# Fix extra spaces and capitalize names properly
df_clean["Customer_Name"] = df_clean["Customer_Name"].str.strip().str.title()

# Convert all emails to lowercase
df_clean["Email"] = df_clean["Email"].str.strip().str.lower()

# Fill missing cities and capitalize properly
df_clean["City"] = df_clean["City"].fillna("Not Specified").str.strip().str.title()

# Remove duplicate customers based on Email
df_clean = df_clean.drop_duplicates(subset=["Email"])

# 3. Export clean data to Excel (CSV)
df_clean.to_csv("cleaned_customer_data.csv", index=False)

print("--- AFTER CLEANING (Ready for Client) ---")
print(df_clean)
print("\nSuccess! Clean data saved to 'cleaned_customer_data.csv'.")
```
