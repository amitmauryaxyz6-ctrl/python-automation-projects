```python

import pandas as pd
from datetime import datetime

# 1. Shortlisted candidates data from HR
shortlisted_candidates = [
    {"Name": "Rohan Gupta", "Email": "rohan@example.com", "Role": "Python Automation Developer", "Interview_Date": "20 Nov 2026", "Time": "11:00 AM"},
    {"Name": "Aarav Sharma", "Email": "aarav@example.com", "Role": "Python Automation Developer", "Interview_Date": "20 Nov 2026", "Time": "02:30 PM"},
    {"Name": "Priya Patel", "Email": "priya@example.com", "Role": "Data Analyst", "Interview_Date": "21 Nov 2026", "Time": "10:30 AM"}
]

company_name = "TechNova Solutions"
hr_name = "HR Recruitment Team"
generated_logs = []

print(f"--- GENERATING AUTOMATED INTERVIEW INVITES ({company_name}) ---\n")

# 2. Generate personalized email & save individual invite file for each candidate
for person in shortlisted_candidates:
    name = person["Name"]
    role = person["Role"]
    date = person["Interview_Date"]
    time = person["Time"]

    email_body = f"""Subject: Interview Invitation for {role} at {company_name}

Dear {name},

Congratulations! Your profile has been shortlisted for the {role} position at {company_name}.

We would like to invite you for an online technical interview:
- Date: {date}
- Time: {time}
- Mode: Google Meet

Please reply to confirm your availability.

Best regards,
{hr_name}
{company_name}"""

    # Save each invite as a separate text file
    file_name = f"Invite_{name.replace(' ', '_')}.txt"
    with open(file_name, "w") as file:
        file.write(email_body)

    generated_logs.append({"Candidate": name, "Email": person["Email"], "File_Created": file_name, "Status": "Ready to Send"})
    print(email_body)
    print("-" * 55)

# 3. Export HR summary log to Excel (CSV)
df_log = pd.DataFrame(generated_logs)
df_log.to_csv("hr_invite_logs.csv", index=False)
print("\nSuccess! All interview letters created and logged in 'hr_invite_logs.csv'.")
```
